```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.optimize import minimize

df = pd.read_csv("./all_redshifts_PVs.csv")
z, mu, err = df.zHD.values, df.MU_SH0ES.values, df.MU_SH0ES_ERR_DIAG.values
m = (z > 0.01) & (err > 0)           # drop very low z (peculiar velocities)
z, mu, err = z[m], mu[m], err[m]
c = 299792.458
```

```python
print(df.columns.tolist())
print(df.head())
```

```python
def dL(z, Om, OL, H0=70):
    Ok = 1 - Om - OL
    E = lambda x: np.sqrt(Om*(1+x)**3 + Ok*(1+x)**2 + OL)
    Dc = c/H0 * np.array([quad(lambda x: 1/E(x), 0, zi)[0] for zi in np.atleast_1d(z)])
    if Ok > 1e-8:   Dm = c/H0/np.sqrt(Ok)*np.sinh(np.sqrt(Ok)*Dc*H0/c)
    elif Ok < -1e-8: Dm = c/H0/np.sqrt(-Ok)*np.sin(np.sqrt(-Ok)*Dc*H0/c)
    else: Dm = Dc
    return (1+z)*Dm

def mu_model(z, Om, OL): return 5*np.log10(dL(z, Om, OL)) + 25

# Fit with a free offset (absorbs M and H0 degeneracy) -> analytic marginalization
def chi2(p):
    Om, OL = p
    r = mu - mu_model(z, Om, OL)
    w = 1/err**2
    off = np.sum(w*r)/np.sum(w)
    return np.sum(w*(r-off)**2)

best = minimize(chi2, [0.3, 0.7], method="Nelder-Mead")
print("Best Om, OL:", best.x, " q0 = Om/2 - OL =", best.x[0]/2 - best.x[1])

# Plot 1: Hubble diagram with models
zz = np.linspace(0.01, 2.2, 80)
off = lambda Om, OL: np.average(mu - mu_model(z, Om, OL), weights=1/err**2)
plt.figure(figsize=(7,5))
plt.errorbar(z, mu, err, fmt='.', alpha=.3, label="Pantheon+")
for Om, OL, lab in [(1,0,"Matter only (EdS)"), (0.3,0.7,"ΛCDM"), (0,0,"Empty")]:
    plt.plot(zz, mu_model(zz, Om, OL) + off(Om, OL), label=lab)
plt.xscale("log"); plt.xlabel("Redshift z"); plt.ylabel("Distance modulus μ"); plt.legend()
plt.savefig("hubble_diagram.png", dpi=200)

# Plot 2: residuals relative to empty universe (the key "acceleration" plot)
plt.figure(figsize=(7,5))
res = mu - (mu_model(z,0,0) + off(0,0))
plt.errorbar(z, res, err, fmt='.', alpha=.2)
plt.plot(zz, mu_model(zz,1,0)+off(1,0)-mu_model(zz,0,0)-off(0,0), label="Matter only")
plt.plot(zz, mu_model(zz,.3,.7)+off(.3,.7)-mu_model(zz,0,0)-off(0,0), label="ΛCDM")
plt.xscale("log"); plt.axhline(0,c='k',lw=.5); plt.legend()
plt.xlabel("z"); plt.ylabel("μ − μ_empty"); plt.savefig("residuals.png", dpi=200)

# Plot 3: low-z linear Hubble law
lo = z < 0.1
d = 10**((mu[lo]-25)/5)
plt.figure(); plt.scatter(c*z[lo], d, s=4)
plt.xlabel("cz (km/s)"); plt.ylabel("d_L (Mpc)")
print("Slope fit H0 ≈", c*z[lo] @ d / (d @ d) if False else 1/np.polyfit(d, c*z[lo], 1)[0]*1)
```
